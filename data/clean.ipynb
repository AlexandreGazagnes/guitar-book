```python
import pandas as pd 
```

```python
df = pd.read_csv('base.csv')
df.head()
```

```python
df.columns
```

```python
for i, v in df.iterrows():
    if isinstance(v['_author'], str):
        if v['_author']:
            continue

    if isinstance(v['_song'], str):
        if v["_song"]:
            continue

    if not isinstance(v["url"], str):
        continue

    if "boite" in v["url"] : 
        print(v["url"])
        # print(v['_author'])
        # print(v['_title'])
        # print(v['_content'])

        _author = v["url"].split("/")[-2].replace("-", " ").title()
        _song = v["url"].split("/")[-1].replace("-", " ").title()

        df.loc[i, "_author"] = _author
        df.loc[i, "_song"] = _song
```

```python
df.to_csv('base.csv', index=False)
```
